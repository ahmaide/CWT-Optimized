```python
import glob
import os
import numpy as np
import matplotlib.pyplot as plt

result_dir = "cwt_results"
result_files = sorted(glob.glob(os.path.join(result_dir, "*.npy")))

print(f"Found {len(result_files)} result files in: {result_dir}/")
print(f"Plotting all...\n")

for i, path in enumerate(result_files):
    name = os.path.basename(path)
    
    coeff = np.load(path)
    
    mag = np.abs(coeff).astype(np.float32)
    viz = np.log1p(mag)
    vmin, vmax = np.percentile(viz, (1, 99))
    
    plt.figure(figsize=(12, 6))
    plt.imshow(viz, origin="lower", aspect="auto", vmin=vmin, vmax=vmax)
    plt.colorbar(label="log(1+|coeff|)")
    plt.title(f"{name} — σ=32")
    plt.xlabel("Time")
    plt.ylabel("Channel")
    plt.tight_layout()
    plt.show()
    
    del coeff, mag, viz
    
    print(f"  {i+1}/{len(result_files)} plotted: {name}")
```
